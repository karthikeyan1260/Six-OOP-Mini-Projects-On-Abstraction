<h1>NOTIFICATION SERVICE USING ABSTRACTION</h1>

```python
from abc import ABC, abstractmethod

# Abstract Class
class Notifier(ABC):

    @abstractmethod
    def send(self, recipient, message):
        pass


# Email Notification
class Email(Notifier):

    def send(self, recipient, message):
        print(f"Email sent to {recipient}:")
        print(message)


# SMS Notification
class SMS(Notifier):

    def send(self, recipient, message):
        print(f"\nSMS sent to {recipient}:")
        print(message)


# Push Notification
class PushNotification(Notifier):

    def send(self, recipient, message):
        print(f"\nPush notification sent to device {recipient}:")
        print(message)


# Creating Objects
email = Email()
sms = SMS()
push = PushNotification()

# Common Message
message = "Your order has been shipped."

# Calling send() method
email.send("arun@example.com", message)

sms.send("+91-9876543210", message)

push.send("A1B2C3", message)
```

<h1>DATABASE REPOSITORY USING ABSTRACTION</h1>

```python
from abc import ABC, abstractmethod

# Abstract Class
class UserRepository(ABC):

    @abstractmethod
    def add(self, name):
        pass

    @abstractmethod
    def get(self, name):
        pass

    @abstractmethod
    def delete(self, name):
        pass


# PostgreSQL Implementation
class PostgreSQL(UserRepository):

    def add(self, name):
        print(f"Adding {name} to PostgreSQL")

    def get(self, name):
        print(f"Getting {name} from PostgreSQL")

    def delete(self, name):
        print(f"Deleting {name} from PostgreSQL")


# MongoDB Implementation
class MongoDB(UserRepository):

    def add(self, name):
        print(f"Adding {name} to MongoDB")

    def get(self, name):
        print(f"Getting {name} from MongoDB")

    def delete(self, name):
        print(f"Deleting {name} from MongoDB")


# Creating Objects
postgres = PostgreSQL()
mongo = MongoDB()

# Common Operations
postgres.add("Priya")
mongo.add("Priya")

postgres.get("Priya")
mongo.get("Priya")

postgres.delete("Priya")
mongo.delete("Priya")
```

<h1>FILE SYSTEM STORAGE USING ABSTRACTION</h1>

```python
from abc import ABC, abstractmethod

# Abstract Class
class FileStorage(ABC):

    @abstractmethod
    def upload(self, filename):
        pass

    @abstractmethod
    def download(self, filename):
        pass

    @abstractmethod
    def delete(self, filename):
        pass


# Local Storage
class LocalStorage(FileStorage):

    def upload(self, filename):
        print(f"Saved {filename} locally")

    def download(self, filename):
        print(f"Downloaded {filename} from Local Storage")

    def delete(self, filename):
        print(f"Deleted {filename} from Local Storage")


# AWS S3 Storage
class AWSS3(FileStorage):

    def upload(self, filename):
        print(f"Uploaded {filename} to AWS S3")

    def download(self, filename):
        print(f"Downloaded {filename} from AWS S3")

    def delete(self, filename):
        print(f"Deleted {filename} from AWS S3")


# Google Cloud Storage
class GoogleCloud(FileStorage):

    def upload(self, filename):
        print(f"Uploaded {filename} to Google Cloud Storage")

    def download(self, filename):
        print(f"Downloaded {filename} from Google Cloud Storage")

    def delete(self, filename):
        print(f"Deleted {filename} from Google Cloud Storage")


# Azure Blob Storage
class AzureBlob(FileStorage):

    def upload(self, filename):
        print(f"Uploaded {filename} to Azure Blob Storage")

    def download(self, filename):
        print(f"Downloaded {filename} from Azure Blob Storage")

    def delete(self, filename):
        print(f"Deleted {filename} from Azure Blob Storage")


# Creating Objects
local = LocalStorage()
aws = AWSS3()
google = GoogleCloud()
azure = AzureBlob()


# Common File Operations
local.upload("resume.pdf")
aws.upload("resume.pdf")

google.download("resume.pdf")
azure.delete("resume.pdf")
```

<h1>VEHICLE SYSTEM USING ABSTRACTION</h1>

```python
from abc import ABC, abstractmethod

# Abstract Class
class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

    # Normal Method
    def vehicle_info(self):
        print(f"This is a {self.__class__.__name__}")


# Car Class
class Car(Vehicle):

    def start(self):
        print("Car engine started with a key")

    def stop(self):
        print("Car engine stopped")


# Bike Class
class Bike(Vehicle):

    def start(self):
        print("Bike engine started with a button")

    def stop(self):
        print("Bike engine stopped")


# Electric Scooter Class
class ElectricScooter(Vehicle):

    def start(self):
        print("Electric scooter powered on")

    def stop(self):
        print("Electric scooter powered off")


# Bus Class
class Bus(Vehicle):

    def start(self):
        print("Bus engine started")

    def stop(self):
        print("Bus engine stopped")


# Creating Objects
car = Car()
bike = Bike()
scooter = ElectricScooter()
bus = Bus()

# Storing Objects in a List
vehicles = [car, bike, scooter, bus]

# Calling Methods
for vehicle in vehicles:
    vehicle.vehicle_info()
    vehicle.start()
    vehicle.stop()
    print()
```

<h1>REPORT GENERATOR USING ABSTRACTION</h1>

```python
from abc import ABC, abstractmethod

# Abstract Class
class ReportExporter(ABC):

    @abstractmethod
    def export(self, data):
        pass


# CSV Exporter
class CSVExporter(ReportExporter):

    def export(self, data):
        print("Exporting CSV Report:")
        print("id,name,city")


# JSON Exporter
class JSONExporter(ReportExporter):

    def export(self, data):
        print("\nExporting JSON Report:")
        print(data)


# Excel Exporter
class ExcelExporter(ReportExporter):

    def export(self, data):
        print("\nExporting Excel Report:")
        print("Excel file generated successfully")


# PDF Exporter
class PDFExporter(ReportExporter):

    def export(self, data):
        print("\nExporting PDF Report:")
        print("PDF file generated successfully")


# Sample Data
data = [
    {"id": 1, "name": "Priya", "city": "Chennai"},
    {"id": 2, "name": "Arun", "city": "Bangalore"}
]

# Creating Objects
csv = CSVExporter()
json = JSONExporter()
excel = ExcelExporter()
pdf = PDFExporter()

# Storing Objects in a List
exporters = [csv, json, excel, pdf]

# Calling Common Method
for exporter in exporters:
    exporter.export(data)
```

<h1>PAYMENT GATEWAY USING ABSTRACTION</h1>

```python
from abc import ABC, abstractmethod

# Abstract Class
class PaymentGateway(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

    @abstractmethod
    def refund(self, transaction_id):
        pass


# Child Class 1
class UPIPayment(PaymentGateway):

    def pay(self, amount):
        return f"Paid Rs.{amount} through UPI"

    def refund(self, transaction_id):
        return f"UPI refund initiated for {transaction_id}"


# Child Class 2
class CardPayment(PaymentGateway):

    def pay(self, amount):
        return f"Paid Rs.{amount} using Card"

    def refund(self, transaction_id):
        return f"Card refund initiated for {transaction_id}"


# Child Class 3
class PayPalPayment(PaymentGateway):

    def pay(self, amount):
        return f"Paid ${amount} using PayPal"

    def refund(self, transaction_id):
        return f"PayPal refund initiated for {transaction_id}"


# Common Function
def checkout(gateway, amount):
    print(gateway.pay(amount))


# Creating Objects
upi = UPIPayment()
card = CardPayment()
paypal = PayPalPayment()

# Payment Processing
checkout(upi, 999)
checkout(card, 2499)
checkout(paypal, 49.99)

# Refund Processing
print()
print(upi.refund("UPI101"))
print(card.refund("CARD102"))
print(paypal.refund("PAYPAL103"))
```
